```python
import numpy as np

# 5 houses, 2 features: [sqft (in 1000s), bath count]
X = np.array([
    [1.0, 1],
    [1.5, 2],
    [2.0, 2],
    [2.5, 3],
    [3.0, 3]
])


y = np.array([20*1.0 + 10*1 + 5,
              20*1.5 + 10*2 + 5,
              20*2.0 + 10*2 + 5,
              20*2.5 + 10*3 + 5,
              20*3.0 + 10*3 + 5])

def compute_cost(X, y, w, b):
    m = len(y)
    predictions = X @ w + b
    errors = predictions - y
    cost = np.sum(errors ** 2) / (2 * m)
    return cost

def compute_gradients(X, y, w, b):
    m = len(y)
    predictions = X @ w + b
    errors = predictions - y
    dw = (X.T @ errors) / m
    db = np.sum(errors) / m
    return dw, db
```

```python
mean = X.mean(axis=0)
std = X.std(axis=0)

X_scaled = (X - mean) / std
X_scaled
```

```python
w = np.zeros(X_scaled.shape[1])
b = 0.0
learning_rate = 0.1
iterations = 1000

cost_history = []

for i in range(iterations):
    dw, db = compute_gradients(X_scaled, y, w, b)
    w = w - learning_rate * dw
    b = b - learning_rate * db

    cost = compute_cost(X_scaled, y, w, b)
    cost_history.append(cost)

    if i % 100 == 0:
        print(f"Iteration {i}: cost = {cost:.4f}, w = {w}, b = {b:.4f}")

print("\nFinal w:", w)
print("Final b:", b)
```
